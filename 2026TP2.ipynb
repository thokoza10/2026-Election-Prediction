```python
pr_2016_ward = (
    df_2016_pr_common
    .groupby(["Ward", "PartyName"], as_index=False)["TotalValidVotes"]
    .sum()
)

pr_2021_ward = (
    df_2021_pr_common
    .groupby(["Ward", "PartyName"], as_index=False)["TotalValidVotes"]
    .sum()
)

print("2016 PR ward-party rows:", len(pr_2016_ward))
print("2021 PR ward-party rows:", len(pr_2021_ward))

print("\n2016 sample:")
print(pr_2016_ward.head())

print("\n2021 sample:")
print(pr_2021_ward.head())
```

Calculate total PR votes per ward

```python
pr_2016_totals = (
    pr_2016_ward
    .groupby("Ward", as_index=False)["TotalValidVotes"]
    .sum()
    .rename(columns={"TotalValidVotes": "WardTotalVotes"})
)

pr_2021_totals = (
    pr_2021_ward
    .groupby("Ward", as_index=False)["TotalValidVotes"]
    .sum()
    .rename(columns={"TotalValidVotes": "WardTotalVotes"})
)

print("2016 wards:", len(pr_2016_totals))
print("2021 wards:", len(pr_2021_totals))

print("\n2016 sample:")
print(pr_2016_totals.head())

print("\n2021 sample:")
print(pr_2021_totals.head())
```

Calculate PR vote share

```python
pr_2016_ward = pr_2016_ward.merge(
    pr_2016_totals,
    on="Ward",
    how="left"
)

pr_2021_ward = pr_2021_ward.merge(
    pr_2021_totals,
    on="Ward",
    how="left"
)

pr_2016_ward["VoteShare"] = (
    pr_2016_ward["TotalValidVotes"] /
    pr_2016_ward["WardTotalVotes"]
) * 100

pr_2021_ward["VoteShare"] = (
    pr_2021_ward["TotalValidVotes"] /
    pr_2021_ward["WardTotalVotes"]
) * 100
```

Calculation of mean, median, variance and range of ward total votes

```python
for year, data in [
    (2016, pr_2016_totals),
    (2021, pr_2021_totals)
]:
    print(f"{year} PR Ward Total Vote Statistics")
    print("Mean:", data["WardTotalVotes"].mean())
    print("Median:", data["WardTotalVotes"].median())
    print("Variance:", data["WardTotalVotes"].var())
    print("Standard Deviation:", data["WardTotalVotes"].std())
    print("Minimum:", data["WardTotalVotes"].min())
    print("Maximum:", data["WardTotalVotes"].max())
    print("Range:", data["WardTotalVotes"].max() - data["WardTotalVotes"].min())
    print()
```

Calculation of the percentage change

```python
mean_2016 = pr_2016_totals["WardTotalVotes"].mean()
mean_2021 = pr_2021_totals["WardTotalVotes"].mean()

percentage_change = (
    (mean_2021 - mean_2016) / mean_2016
) * 100

print("Percentage change in mean PR votes per ward:")
print(round(percentage_change, 2), "%")
```

party-level mathematical analysis

```python
party_totals_2016 = (
    pr_2016_ward
    .groupby("PartyName")["TotalValidVotes"]
    .sum()
    .sort_values(ascending=False)
)

party_totals_2021 = (
    pr_2021_ward
    .groupby("PartyName")["TotalValidVotes"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 parties in 2016:")
print(party_totals_2016.head(10))

print("\nTop 10 parties in 2021:")
print(party_totals_2021.head(10))
```

percentage change by major party

```python
major_parties = [
    "AFRICAN NATIONAL CONGRESS",
    "DEMOCRATIC ALLIANCE",
    "ECONOMIC FREEDOM FIGHTERS",
    "INKATHA FREEDOM PARTY",
    "ACTIONSA"
]

party_change = pd.DataFrame({
    "PartyName": major_parties,
    "2016Votes": [
        party_totals_2016.get(party, 0) for party in major_parties
    ],
    "2021Votes": [
        party_totals_2021.get(party, 0) for party in major_parties
    ]
})

party_change["VoteChange"] = (
    party_change["2021Votes"] - party_change["2016Votes"]
)

party_change["PercentageChange"] = (
    party_change["VoteChange"] /
    party_change["2016Votes"].replace(0, pd.NA)
) * 100

print(party_change)
```

vote-share analysis

```python
total_votes_2016 = party_totals_2016.sum()
total_votes_2021 = party_totals_2021.sum()

party_share = party_change.copy()

party_share["2016Share"] = (
    party_share["2016Votes"] / total_votes_2016
) * 100

party_share["2021Share"] = (
    party_share["2021Votes"] / total_votes_2021
) * 100

party_share["ShareChange"] = (
    party_share["2021Share"] - party_share["2016Share"]
)

print(party_share)
```

Train the model

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    max_depth=10,
    min_samples_leaf=2
)

model.fit(X_train, y_train)

print("Model training completed successfully.")
```

Prepare turnout data

```python
turnout_2016 = (
    df_2016_pr
    .groupby(
        ["Ward", "VotingDistrict", "VotingStationName"],
        as_index=False
    )
    .agg(
        RegisteredVoters=("RegisteredVoters", "first"),
        SpoiltVotes=("SpoiltVotes", "first"),
        ValidVotes=("TotalValidVotes", "sum")
    )
)

turnout_2021 = (
    df_2021_pr
    .groupby(
        ["Ward", "VotingDistrict", "VotingStationName"],
        as_index=False
    )
    .agg(
        RegisteredVoters=("RegisteredVoters", "first"),
        SpoiltVotes=("SpoiltVotes", "first"),
        ValidVotes=("TotalValidVotes", "sum")
    )
)

print("2016 turnout records:", turnout_2016.shape)
print("2021 turnout records:", turnout_2021.shape)

print("\n2016 sample:")
print(turnout_2016.head())

print("\n2021 sample:")
print(turnout_2021.head())
```

Calculate turnout rates

```python
turnout_2016["TurnoutRate"] = (
    (turnout_2016["ValidVotes"] + turnout_2016["SpoiltVotes"])
    / turnout_2016["RegisteredVoters"]
) * 100

turnout_2021["TurnoutRate"] = (
    (turnout_2021["ValidVotes"] + turnout_2021["SpoiltVotes"])
    / turnout_2021["RegisteredVoters"]
) * 100
```

aggregation of the stations to wards:

```python
ward_turnout_2016 = (
    turnout_2016
    .groupby("Ward", as_index=False)
    .agg(
        RegisteredVoters_2016=("RegisteredVoters", "sum"),
        ValidVotes_2016=("ValidVotes", "sum"),
        SpoiltVotes_2016=("SpoiltVotes", "sum")
    )
)

ward_turnout_2021 = (
    turnout_2021
    .groupby("Ward", as_index=False)
    .agg(
        RegisteredVoters_2021=("RegisteredVoters", "sum"),
        ValidVotes_2021=("ValidVotes", "sum"),
        SpoiltVotes_2021=("SpoiltVotes", "sum")
    )
)

ward_turnout_2016["TurnoutRate_2016"] = (
    (
        ward_turnout_2016["ValidVotes_2016"]
        + ward_turnout_2016["SpoiltVotes_2016"]
    )
    / ward_turnout_2016["RegisteredVoters_2016"]
) * 100

ward_turnout_2021["TurnoutRate_2021"] = (
    (
        ward_turnout_2021["ValidVotes_2021"]
        + ward_turnout_2021["SpoiltVotes_2021"]
    )
    / ward_turnout_2021["RegisteredVoters_2021"]
) * 100
```

compare the two years:

```python
turnout_comparison = ward_turnout_2016.merge(
    ward_turnout_2021,
    on="Ward",
    how="inner"
)

turnout_comparison["TurnoutChange"] = (
    turnout_comparison["TurnoutRate_2021"]
    - turnout_comparison["TurnoutRate_2016"]
)

print("Turnout comparison:")
print(turnout_comparison.head())

print("\nAverage 2016 turnout:",
      turnout_comparison["TurnoutRate_2016"].mean())

print("Average 2021 turnout:",
      turnout_comparison["TurnoutRate_2021"].mean())

print("Average turnout change:",
      turnout_comparison["TurnoutChange"].mean())
```

derived 2026 projection,

```python
average_turnout_2016 = turnout_comparison["TurnoutRate_2016"].mean()
average_turnout_2021 = turnout_comparison["TurnoutRate_2021"].mean()

turnout_ratio = (
    average_turnout_2021 / average_turnout_2016
)

projected_turnout_2026 = (
    average_turnout_2021 * turnout_ratio
)

print("Average 2016 PR participation:",
      round(average_turnout_2016, 2), "%")

print("Average 2021 PR participation:",
      round(average_turnout_2021, 2), "%")

print("Historical turnout ratio:",
      round(turnout_ratio, 4))

print("Projected 2026 PR participation:",
      round(projected_turnout_2026, 2), "%")
```

Estimate the number of 2026 voters

```python
total_registered_2021 = (
    turnout_comparison["RegisteredVoters_2021"].sum()
)

estimated_voters_2026 = (
    total_registered_2021
    * (projected_turnout_2026 / 100)
)

print("Total registered voters in 2021:",
      total_registered_2021)

print("Projected 2026 PR participation:",
      round(projected_turnout_2026, 2), "%")

print("Estimated 2026 participating voters:",
      round(estimated_voters_2026))
```

```python
overall_2026["ProjectedVotes_2026"] = (
    overall_2026["AveragePredictedVoteShare_2026"]
    / overall_2026["AveragePredictedVoteShare_2026"].sum()
) * estimated_voters_2026

overall_2026["ProjectedVotes_2026"] = (
    overall_2026["ProjectedVotes_2026"].round().astype(int)
)

overall_2026["ProjectedShare_2026"] = (
    overall_2026["ProjectedVotes_2026"]
    / overall_2026["ProjectedVotes_2026"].sum()
) * 100

print("2026 projected party results:")
print(overall_2026.head(15))
```
